```python
from itertools import batched

import torch
import torch.nn.functional as F

from gpt2.models.gpt2_model import GPT2Model, GPT2Tokenizer
from gpt2.train import AIAYNScheduler
from datasets import load_dataset
from torch.utils.data import DataLoader
```

```python
def format_pair(record: dict) -> dict:
    """
    Format each english-spanish pair as <en>text</en><es>text</es>.
    """
    en = record["translation"]["en"]
    es = record["translation"]["es"]
    record["translation"] = f"<en>{en}</en><es>{es}</es>"
    return record


def tokenize(tokenizer: GPT2Tokenizer, record: dict) -> dict:
    record["tokenized"] = torch.concat((tokenizer.encode(record["translation"]), torch.tensor([tokenizer.end_of_text_token()])))
    return record


def create_minibatches(examples, seq_len):
    concatenated = torch.concat(examples["tokenized"])
    max_upper = seq_len * (concatenated.shape[-1] // seq_len)
    # TODO: The GPT-3 paper actually randomly samples 512-token sequences from the dataset. Replicate that here.
    tokens = torch.reshape(concatenated[:max_upper], (-1, seq_len))
    return {"tokens": tokens}
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = GPT2Model(d_model=768, d_ff=3072, n_heads=12, n_heads=12, max_context_length=512, p_dropout=0.1).to(device)
```

```python
if torch.cuda.is_available():
    torch.cuda.empty_cache()
```

```python
# Let's pretrain on some a translation dataset again
split = "train[:10_000]"
dataset = load_dataset("Helsinki-NLP/opus-100", "en-es", split=split).with_format("torch")
transformed_dataset = dataset.map(format_pair, desc="format")
transformed_dataset = transformed_dataset.map(lambda example: tokenize(model.tokenizer, example), desc="tokenize")
transformed_dataset = transformed_dataset.map(lambda examples: create_minibatches(examples, 512), batched=True, batch_size=None, remove_columns=["tokenized", "translation"], desc="chunk")
```

```python
data_loader = DataLoader(transformed_dataset, batch_size=32, num_workers=4, pin_memory=True)
optim = torch.optim.Adam(model.parameters(), lr=3e-4, betas=(0.9, 0.98), eps=1e-9)
scheduler = AIAYNScheduler(optim, model.d_model)

model.train()
```

```python
for epoch in range(100):
    for i, batch in enumerate(data_loader):
        optim.zero_grad()

        tokens = batch["tokens"].to(device, non_blocking=True)
        logits = model.forward(tokens)
        
        # reshape the preds to a tensor of [all_tokens, vocab_size], which we'll then pair with the targets, which is [all_tokens]
        preds = logits.view(-1, model.tokenizer.vocab_size())
        targets = tokens.view(-1)
        
        loss = F.cross_entropy(preds, targets)
        loss.backward()
        optim.step()
        scheduler.step()
        if i % 10 == 0:
            print(f"train loss: {loss.item()}")
```
